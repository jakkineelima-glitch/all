```python
# ============================================================
# SCREEN ROTATION AS ORTHOGONAL TRANSFORMATION (F6)
# B.Tech 1st-Year Mathematics Mini Project
# ============================================================

"""
Problem Statement:
Phone screen rotation can be mathematically represented using
an orthogonal matrix.

We will:
1. Construct the rotation matrix.
2. Show that Q^T Q = I.
3. Verify det(Q) = 1.
4. Apply rotation to screen coordinates.
5. Show that distances are preserved.
6. Show quadratic form invariance.
7. Show that rotating twice gives a 180-degree rotation.
8. Visualize the transformation using graphs.
"""

print("SCREEN ROTATION AS ORTHOGONAL TRANSFORMATION")
print("=" * 55)
print("Mathematical model: 2D Rotation Matrix")
print("Environment: Google Colab + Python")
```

```python
# ============================================================
# IMPORTS
# ============================================================

import numpy as np
import sympy as sp
import matplotlib.pyplot as plt

from IPython.display import display, Math

# Make NumPy output easier to read
np.set_printoptions(precision=4, suppress=True)

print("Libraries imported successfully.")
```

```python
# ============================================================
# USER CONFIGURATION
# ============================================================

# Rotation angle in degrees
ANGLE_DEGREES = 90

# A sample screen coordinate.
# Coordinates are measured relative to the screen center.
POINT = np.array([3.0, 2.0])

# A second point for checking distance preservation
POINT_2 = np.array([1.0, -1.0])

# Number of times the rotation is applied
NUMBER_OF_ROTATIONS = 2

print("Input Configuration")
print("-" * 30)
print(f"Rotation angle       : {ANGLE_DEGREES}°")
print(f"Point P              : {POINT}")
print(f"Point R              : {POINT_2}")
print(f"Number of rotations  : {NUMBER_OF_ROTATIONS}")
```

```python
# ============================================================
# MATHEMATICAL MODEL
# ============================================================

theta = sp.symbols('theta', real=True)

Q = sp.Matrix([
    [sp.cos(theta), -sp.sin(theta)],
    [sp.sin(theta),  sp.cos(theta)]
])

print("General 2D Rotation Matrix:")
display(Q)

print("\nMathematical form:")
display(
    Math(
        r"Q(\theta)=\begin{bmatrix}"
        r"\cos\theta & -\sin\theta\\"
        r"\sin\theta & \cos\theta"
        r"\end{bmatrix}"
    )
)
```

```python
# ============================================================
# ROTATION MATRIX FOR USER'S ANGLE
# ============================================================

theta_value = sp.pi * ANGLE_DEGREES / 180

Q_angle = sp.simplify(Q.subs(theta, theta_value))

print(f"Rotation Matrix for {ANGLE_DEGREES}°:")
display(Q_angle)

print("\nNumerical Rotation Matrix:")
Q_numeric = np.array(Q_angle.evalf(), dtype=float)
print(Q_numeric)
```

```python
# ============================================================
# ORTHOGONALITY CHECK
# ============================================================

Q_transpose = Q_angle.T
QTQ = sp.simplify(Q_transpose * Q_angle)

identity_matrix = sp.eye(2)

print("Q:")
display(Q_angle)

print("\nQ^T:")
display(Q_transpose)

print("\nQ^T Q:")
display(QTQ)

print("\nIdentity Matrix I:")
display(identity_matrix)

is_orthogonal = QTQ == identity_matrix

print("\nOrthogonality Test:")
print("Q^T Q = I ?", is_orthogonal)
```

```python
# ============================================================
# DETERMINANT
# ============================================================

det_Q = sp.simplify(Q_angle.det())

print("Determinant of Q:")
display(det_Q)

print("\nNumerical determinant:", float(det_Q))

if abs(float(det_Q) - 1) < 1e-10:
    print("Result: det(Q) = +1")
    print("This represents a proper rotation.")
elif abs(float(det_Q) + 1) < 1e-10:
    print("Result: det(Q) = -1")
    print("This represents an orthogonal transformation involving reflection.")
```

```python
# ============================================================
# APPLY ROTATION TO A SCREEN POINT
# ============================================================

P = sp.Matrix(POINT)

P_rotated = Q_angle * P

print("Original Point P:")
display(P)

print("\nRotated Point QP:")
display(P_rotated)

P_rotated_numeric = np.array(P_rotated.evalf(), dtype=float).flatten()

print("\nNumerical Result:")
print(f"P  = {POINT}")
print(f"QP = {P_rotated_numeric}")
```

```python
# ============================================================
# DISTANCE / LENGTH PRESERVATION
# ============================================================

P_np = POINT
P_rot_np = P_rotated_numeric

original_length = np.linalg.norm(P_np)
rotated_length = np.linalg.norm(P_rot_np)

print("Length Preservation")
print("-" * 35)

print(f"Original length : {original_length:.6f}")
print(f"Rotated length  : {rotated_length:.6f}")

print(f"\nDifference      : {abs(original_length - rotated_length):.10f}")

if np.isclose(original_length, rotated_length):
    print("\n✓ Distance/length is preserved.")
else:
    print("\n✗ Distance/length is not preserved.")
```

```python
# ============================================================
# DISTANCE BETWEEN TWO POINTS
# ============================================================

R_np = POINT_2

P_rot = Q_numeric @ P_np
R_rot = Q_numeric @ R_np

original_distance = np.linalg.norm(P_np - R_np)
rotated_distance = np.linalg.norm(P_rot - R_rot)

print("Distance Between Two Screen Coordinates")
print("-" * 45)

print(f"Original distance : {original_distance:.6f}")
print(f"Rotated distance  : {rotated_distance:.6f}")

print(f"Difference        : {abs(original_distance - rotated_distance):.10f}")

if np.isclose(original_distance, rotated_distance):
    print("\n✓ Distance is preserved by the rotation.")
```

```python
⅞# ============================================================
# QUADRATIC FORM INVARIANCE
# ============================================================

x1, x2 = sp.symbols('x1 x2', real=True)

x = sp.Matrix([x1, x2])

original_quadratic = sp.expand((x.T * x)[0])

transformed_x = Q * x

rotated_quadratic = sp.simplify(
    (transformed_x.T * transformed_x)[0]
)

print("Original Quadratic Form:")
display(original_quadratic)

print("\nAfter Rotation:")
display(rotated_quadratic)

print("\nDifference:")
display(sp.simplify(rotated_quadratic - original_quadratic))

print("\nTherefore:")
print("(Qx)^T(Qx) = x^T x")
print("The quadratic form is invariant under orthogonal rotation.")
```

```python
# ============================================================
# NUMERICAL QUADRATIC FORM TEST
# ============================================================

x_original = POINT
x_rotated = Q_numeric @ x_original

quadratic_original = x_original @ x_original
quadratic_rotated = x_rotated @ x_rotated

print("Numerical Quadratic Form Test")
print("-" * 40)

print(f"x^T x before rotation : {quadratic_original:.6f}")
print(f"x^T x after rotation  : {quadratic_rotated:.6f}")

if np.isclose(quadratic_original, quadratic_rotated):
    print("\n✓ Quadratic form is invariant.")
else:
    print("\n✗ Quadratic form changed.")
```

```python
# ============================================================
# ROTATING TWICE
# ============================================================

Q_twice = sp.simplify(Q_angle * Q_angle)

print("Rotation Matrix Q:")
display(Q_angle)

print("\nQ × Q:")
display(Q_twice)

Q_180 = sp.Matrix([
    [sp.cos(sp.pi), -sp.sin(sp.pi)],
    [sp.sin(sp.pi),  sp.cos(sp.pi)]
])

Q_180 = sp.simplify(Q_180)

print("\nExpected 180° Rotation Matrix:")
display(Q_180)

print("\nDoes Q² = Q(180°)?")
print(Q_twice == Q_180)
```

```python
# ============================================================
# APPLY TWO ROTATIONS TO THE POINT
# ============================================================

P_after_one = Q_numeric @ POINT
P_after_two = Q_numeric @ P_after_one

print("Repeated Rotation")
print("-" * 35)

print("Original point:")
print(POINT)

print("\nAfter first 90° rotation:")
print(P_after_one)

print("\nAfter second 90° rotation:")
print(P_after_two)

print("\nExpected after 180° rotation:")
print(-POINT)

if np.allclose(P_after_two, -POINT):
    print("\n✓ Two 90° rotations = one 180° rotation.")
```

```python
# ============================================================
# GENERAL ROTATION FUNCTION
# ============================================================

def rotation_matrix_degrees(angle_degrees):
    """
    Returns a standard 2D rotation matrix.

    angle_degrees:
        Rotation angle in degrees.
    """

    angle_radians = np.radians(angle_degrees)

    return np.array([
        [np.cos(angle_radians), -np.sin(angle_radians)],
        [np.sin(angle_radians),  np.cos(angle_radians)]
    ])


def rotate_point(point, angle_degrees):
    """
    Rotates a 2D point using the rotation matrix.
    """

    Q = rotation_matrix_degrees(angle_degrees)

    point = np.asarray(point, dtype=float)

    return Q @ point


# Example
test_point = np.array([4.0, 1.0])

rotated_test_point = rotate_point(test_point, 90)

print("Original:", test_point)
print("Rotated :", rotated_test_point)
```

```python
# ============================================================
# AUTOMATIC MATHEMATICAL VERIFICATION
# ============================================================

def verify_rotation(angle_degrees, point):
    Q = rotation_matrix_degrees(angle_degrees)

    # Orthogonality
    orthogonality = Q.T @ Q
    I = np.eye(2)

    # Determinant
    determinant = np.linalg.det(Q)

    # Point transformation
    rotated_point = Q @ point

    # Length preservation
    original_length = np.linalg.norm(point)
    rotated_length = np.linalg.norm(rotated_point)

    # Quadratic form
    original_qf = point @ point
    rotated_qf = rotated_point @ rotated_point

    print("=" * 60)
    print("ROTATION VERIFICATION")
    print("=" * 60)

    print("\nRotation Matrix:")
    print(Q)

    print("\nQ^T Q:")
    print(orthogonality)

    print("\nIs Q orthogonal?")
    print(np.allclose(orthogonality, I))

    print("\nDeterminant:")
    print(determinant)

    print("\nOriginal point:")
    print(point)

    print("\nRotated point:")
    print(rotated_point)

    print("\nOriginal length:")
    print(original_length)

    print("\nRotated length:")
    print(rotated_length)

    print("\nQuadratic form before:")
    print(original_qf)

    print("\nQuadratic form after:")
    print(rotated_qf)

    print("\n" + "=" * 60)

    if (
        np.allclose(orthogonality, I)
        and np.isclose(abs(determinant), 1)
        and np.isclose(original_length, rotated_length)
        and np.isclose(original_qf, rotated_qf)
    ):
        print("✓ ALL MATHEMATICAL TESTS PASSED")
    else:
        print("✗ SOME TESTS FAILED")

    print("=" * 60)


verify_rotation(ANGLE_DEGREES, POINT)
```

```python
# ============================================================
# BENCHMARK TEST DATASET
# ============================================================

benchmark_data = [
    {
        "name": "Portrait → Landscape",
        "angle": 90,
        "point": [3, 2]
    },
    {
        "name": "Landscape → Portrait",
        "angle": -90,
        "point": [4, 1]
    },
    {
        "name": "180 Degree Rotation",
        "angle": 180,
        "point": [2, -3]
    },
    {
        "name": "45 Degree Rotation",
        "angle": 45,
        "point": [1, 1]
    },
    {
        "name": "Full Rotation",
        "angle": 360,
        "point": [5, 2]
    }
]

print("Benchmark Dataset")
print("=" * 60)

for item in benchmark_data:
    print(
        f"{item['name']:25s} | "
        f"Angle = {item['angle']:>4}° | "
        f"Point = {item['point']}"
    )
```

```python
# ============================================================
# BENCHMARK VERIFICATION
# ============================================================

benchmark_results = []

for item in benchmark_data:

    angle = item["angle"]
    point = np.array(item["point"], dtype=float)

    Q = rotation_matrix_degrees(angle)

    rotated = Q @ point

    # Orthogonality
    orthogonal = np.allclose(
        Q.T @ Q,
        np.eye(2)
    )

    # Determinant
    determinant_ok = np.isclose(
        abs(np.linalg.det(Q)),
        1
    )

    # Length preservation
    length_ok = np.isclose(
        np.linalg.norm(point),
        np.linalg.norm(rotated)
    )

    # Quadratic form invariance
    quadratic_ok = np.isclose(
        point @ point,
        rotated @ rotated
    )

    all_passed = (
        orthogonal
        and determinant_ok
        and length_ok
        and quadratic_ok
    )

    benchmark_results.append([
        item["name"],
        angle,
        tuple(point),
        tuple(np.round(rotated, 4)),
        orthogonal,
        determinant_ok,
        length_ok,
        quadratic_ok,
        all_passed
    ])

print("Benchmark Results")
print("=" * 110)

for result in benchmark_results:
    print(
        f"{result[0]:25s} | "
        f"{result[1]:>4}° | "
        f"{str(result[2]):15s} → {str(result[3]):18s} | "
        f"PASS = {result[-1]}"
    )
```

```python
import numpy as np
import sympy as sp
import matplotlib.pyplot as plt

np.set_printoptions(precision=4, suppress=True)

print("Libraries imported successfully!")
```

```python
# Rotation angle
angle_degrees = 90

# Screen coordinate point
point = np.array([3.0, 2.0])

print("Rotation Angle:", angle_degrees, "degrees")
print("Original Point:", point)
```

```python
# Convert degrees to radians
theta = np.radians(angle_degrees)

# Standard 2D rotation matrix
Q = np.array([
    [np.cos(theta), -np.sin(theta)],
    [np.sin(theta),  np.cos(theta)]
])

print("Rotation Matrix Q:")
print(Q)
```

```python
# Rotate the point
rotated_point = Q @ point

print("Original Point:")
print(point)

print("\nRotated Point:")
print(rotated_point)
```

```python
# Transpose of Q
QT = Q.T

# Calculate Q^T Q
QTQ = QT @ Q

# Identity matrix
I = np.eye(2)

print("Q^T Q =")
print(QTQ)

print("\nIdentity Matrix I =")
print(I)

print("\nIs Q an orthogonal matrix?")
print(np.allclose(QTQ, I))
```

```python
det_Q = np.linalg.det(Q)

print("Determinant of Q =", round(det_Q, 4))

if np.isclose(abs(det_Q), 1):
    print("✓ |det(Q)| = 1")
    print("Therefore, Q is an orthogonal transformation.")
```

```python
# Length before rotation
original_length = np.linalg.norm(point)

# Length after rotation
rotated_length = np.linalg.norm(rotated_point)

print("Original length:", original_length)
print("Rotated length :", rotated_length)

if np.isclose(original_length, rotated_length):
    print("\n✓ Distance is preserved.")
else:
    print("\n✗ Distance is not preserved.")
```

```python
# Quadratic form before rotation
original_quadratic = point @ point

# Quadratic form after rotation
rotated_quadratic = rotated_point @ rotated_point

print("Quadratic form before rotation:")
print(original_quadratic)

print("\nQuadratic form after rotation:")
print(rotated_quadratic)

if np.isclose(original_quadratic, rotated_quadratic):
    print("\n✓ Quadratic form is invariant.")
```

```python
(Qx)^T(Qx)
= x^T Q^T Q x
= x^T I x
= x^T x
```

```python
# Apply the rotation a second time
rotated_twice = Q @ rotated_point

# Matrix multiplication Q²
Q_squared = Q @ Q

print("After first rotation:")
print(rotated_point)

print("\nAfter second rotation:")
print(rotated_twice)

print("\nQ² =")
print(Q_squared)

print("\nExpected 180° rotation matrix:")
print(rotation_matrix_180 := np.array([
    [-1, 0],
    [0, -1]
]))

if np.allclose(Q_squared, rotation_matrix_180):
    print("\n✓ Two 90° rotations = one 180° rotation.")
```
